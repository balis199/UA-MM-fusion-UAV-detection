```python
# define path to dataset
from pathlib import Path

dataset_path = Path("../data/Anti-UAV-RGBT")

# define subfolders
subfolders = ["train", "val", "test"]

print(dataset_path / subfolders[0])
```

```python
# import and display a single frame (rgb/ir)
import cv2
import matplotlib.pyplot as plt

seq = dataset_path / "test" / "20190925_111757_1_1"

cap = cv2.VideoCapture(str(seq / "visible.mp4"))
_, rgb = cap.read()
cap.release()

cap = cv2.VideoCapture(str(seq / "infrared.mp4"))
_, ir = cap.read()
cap.release()

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(rgb, cv2.COLOR_BGR2RGB))
plt.title("rgb")
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(ir, cv2.COLOR_BGR2RGB))
plt.title("ir")
plt.show()
```

```python
# project gt label on top of that frame
import json
from typing import Any

gt: Any = json.loads((seq / "visible.json").read_text())
x, y, w, h = gt["gt_rect"][0]

cv2.rectangle(rgb, (x, y), (x + w, y + h), (0, 255, 0), 3)


plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(rgb, cv2.COLOR_BGR2RGB))
plt.title(f"gt_rect = {(x, y, w, h)}")

gt_ir: Any = json.loads((seq / "infrared.json").read_text())
x, y, w, h = gt_ir["gt_rect"][0]

plt.subplot(1, 2, 2)
cv2.rectangle(ir, (x, y), (x + w, y + h), (0, 255, 0), 3)
plt.imshow(cv2.cvtColor(ir, cv2.COLOR_BGR2RGB))
plt.title(f"gt_rect = {(x, y, w, h)}")
plt.show()
```
